```python
import pandas as pd
import numpy as np

from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.preprocessing import StandardScaler

# =========================
# 1. Load Data
# =========================
df = pd.read_csv("Task_execution_observations.csv")  # <-- change filename

# =========================
# 2. Select Variables
# =========================
features = [
    "Prompt_Iterations",
    "Corrective_Prompts",
    "Context_Prompts",
    "Modified_Files",
    "Validation_Detected",
    "Manual_Corrections"
]

target = "Actual_Time_Min"

X = df[features]
y = df[target]

# =========================
# 3. Standardize Features
# =========================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# =========================
# 4. K-Fold Setup
# =========================
k = 5  # you can also try 10
kf = KFold(n_splits=k, shuffle=True, random_state=42)

# =========================
# 5. Model
# =========================
model = LinearRegression()

r2_scores = []
rmse_scores = []
mae_scores = []

# =========================
# 6. Cross Validation Loop
# =========================
for train_index, test_index in kf.split(X_scaled):

    X_train, X_test = X_scaled[train_index], X_scaled[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    # Train
    model.fit(X_train, y_train)

    # Predict
    y_pred = model.predict(X_test)

    # Metrics
    r2_scores.append(r2_score(y_test, y_pred))
    rmse_scores.append(np.sqrt(mean_squared_error(y_test, y_pred)))
    mae_scores.append(mean_absolute_error(y_test, y_pred))

# =========================
# 7. Results
# =========================
print("K-Fold Cross Validation Results (k = {})".format(k))
print("-" * 50)

print(f"R² Mean: {np.mean(r2_scores):.4f}")
print(f"R² Std : {np.std(r2_scores):.4f}")

print(f"RMSE Mean: {np.mean(rmse_scores):.4f}")
print(f"RMSE Std : {np.std(rmse_scores):.4f}")

print(f"MAE Mean: {np.mean(mae_scores):.4f}")
print(f"MAE Std : {np.std(mae_scores):.4f}")
```
